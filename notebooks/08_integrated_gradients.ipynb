```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
import torch
import matplotlib.pyplot as plt

from utils.model import build_resnet50
from utils.dataloaders import create_dataloaders
from utils.checkpoint import load_checkpoint

from utils.explainability import tensor_to_image

from utils.integrated_gradients import (
    generate_integrated_gradients,
    overlay_integrated_gradients
)
```

```python
if torch.backends.mps.is_available():
    device = torch.device("mps")

elif torch.cuda.is_available():
    device = torch.device("cuda")

else:
    device = torch.device("cpu")

print(device)
```

```python
dataset_path = "../datasets/COVID-19_Radiography_Dataset"

(
    train_loader,
    val_loader,
    test_loader,
    train_df,
    val_df,
    test_df
) = create_dataloaders(
    dataset_path=dataset_path,
    batch_size=1
)
```

```python
model = build_resnet50()

model = load_checkpoint(
    model,
    "../checkpoints/best_resnet50.pth",
    device
)

model = model.to(device)
model.eval()
```

```python
class_names = {
    0: "COVID",
    1: "Normal",
    2: "Lung Opacity",
    3: "Viral Pneumonia"
}
```

```python
correct_samples = {}

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)
        prediction = outputs.argmax(dim=1)

        true_label = labels.item()

        if prediction.item() == true_label and true_label not in correct_samples:

            correct_samples[true_label] = (
                images.cpu(),
                labels.cpu()
            )

        if len(correct_samples) == 4:
            break

print(correct_samples.keys())
```

```python
from pathlib import Path

save_dir = Path("../figures/integrated_gradients")
save_dir.mkdir(parents=True, exist_ok=True)

for class_id, (images, labels) in correct_samples.items():

    images = images.to(device)

    with torch.no_grad():

        outputs = model(images)

        probs = torch.softmax(outputs, dim=1)

        confidence, prediction = torch.max(probs,1)

    ig_map = generate_integrated_gradients(
        model,
        images,
        prediction.item()
    )

    original = tensor_to_image(images)

    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)

    plt.imshow(original)

    plt.title(
        f"Original\nTrue: {class_names[class_id]}"
    )

    plt.axis("off")

    plt.subplot(1,2,2)

    overlay = overlay_integrated_gradients(
    original,
    ig_map
)

    plt.imshow(overlay)  
    plt.title(
        f"Integrated Gradients\nPred: {class_names[prediction.item()]}\nConfidence: {confidence.item()*100:.2f}%"
    )

    plt.axis("off")

    plt.tight_layout()

    filename = (
        f"true_{class_names[class_id].replace(' ','_')}"
        f"_pred_{class_names[prediction.item()].replace(' ','_')}.png"
    )

    plt.savefig(
        save_dir / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

print("Integrated Gradients completed.")
```
