```python
baseline = pd.read_csv("../results/baseline_image_results.csv")
iegef = pd.read_csv("../results/iegef_image_results.csv")

comparison = pd.read_csv("../results/baseline_vs_iegef.csv")
```

```python
plt.figure(figsize=(8,6))

data = [

    baseline[baseline.Method=="Grad-CAM"]["IoU"],

    iegef[iegef.Method=="Grad-CAM"]["IoU"],

    baseline[baseline.Method=="Grad-CAM++"]["IoU"],

    iegef[iegef.Method=="Grad-CAM++"]["IoU"]

]

labels = [

    "Baseline\nGradCAM",

    "IEGEF\nGradCAM",

    "Baseline\nGradCAM++",

    "IEGEF\nGradCAM++"

]

plt.boxplot(data, tick_labels=labels)

plt.ylabel("IoU")

plt.title("IoU Comparison")

plt.show()
```
